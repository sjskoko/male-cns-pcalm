# Train on fruit-fly wiring
CPU demonstration of BP, PC and PC-ALM on a real MaleCNS-derived feedforward circuit.

**Not a whole-brain simulation.** 132 neurons, 1,001 retained edges; synthetic regression targets. One seed and 20 epochs illustrate execution, not a scientific comparison. Published study: 760 fits; no demonstrated PC-ALM advantage.

Choose **Runtime → Run all** in Colab. This downloads public code and installs dependencies. No GPU or raw connectome download needed.


```python
import os, sys, subprocess, tempfile
from pathlib import Path
os.environ['OPENBLAS_NUM_THREADS'] = '1'
os.environ['OMP_NUM_THREADS'] = '1'
workspace = Path(tempfile.mkdtemp(prefix='malecns-demo-'))
repo = workspace / 'repo'
subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/sjskoko/male-cns-pcalm.git', str(repo)], check=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', str(repo / 'experiments/credit_rebuild/requirements.txt')], check=True)
print('Code revision:', subprocess.check_output(['git', '-C', str(repo), 'rev-parse', 'HEAD'], text=True).strip())
```

```python
import importlib.util
spec = importlib.util.spec_from_file_location('launch_demo', repo / 'examples/launch_demo.py')
demo = importlib.util.module_from_spec(spec)
spec.loader.exec_module(demo)
results, figure = demo.run_demo()
import matplotlib.pyplot as plt
plt.show()
```

## Interpret carefully
All three methods share the same synthetic data and initialization, optimizer learning rate 0.003 and 20 epochs. PC/PC-ALM use 8 inference steps; PC-ALM dual rate is 0.2. Equal epoch counts do not imply equal compute.

A single seed is not evidence of a general ranking. The full study uses separate development/evaluation seeds and additional matched-rate controls. This demo writes no published experiment records.
